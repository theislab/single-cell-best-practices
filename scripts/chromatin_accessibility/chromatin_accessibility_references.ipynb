# Reference files for the chromatin accessibility chapters

The chromatin accessibility chapters need the gene annotation and genome sequence of GRCh38 and a collection of transcription factor motifs.
We store the files that SnapATAC2 uses by default in lamindb, so that every chapter loads the same versions.
The genome sequence is stored uncompressed, since reading the sequences of many regions from a gzip-compressed file is slow.

```python
import lamindb as ln

ln.connect("theislab/sc-best-practices")

ln.track("K04IXpd6yTL2")
```

```python
from pathlib import Path

import pooch
```

We download the GENCODE release 41 annotation and genome sequence of GRCh38 and the human CIS-BP motifs, checking them against the hashes that SnapATAC2 uses.

```python
gencode = "https://ftp.ebi.ac.uk/pub/databases/gencode/Gencode_human/release_41"
annotation = Path(
    pooch.retrieve(
        f"{gencode}/gencode.v41.basic.annotation.gff3.gz",
        known_hash="sha256:b82a655bdb736ca0e463a8f5d00242bedf10fa88ce9d651a017f135c7c4e9285",
    )
)
fasta = Path(
    pooch.retrieve(
        f"{gencode}/GRCh38.primary_assembly.genome.fa.gz",
        known_hash="sha256:4fac949d7021cbe11117ddab8ec1960004df423d672446cadfbc8cca8007e228",
        processor=pooch.Decompress(name="gencode_v41_GRCh38.fa"),
    )
)
motifs = Path(
    pooch.retrieve(
        "https://exampledata.scverse.org/snapatac2/cisBP_human.meme",
        known_hash="sha256:8bf995450258e61cb1c535d5bf9656d580eb68ba68893fa36b77d17ee0730579",
    )
)
```

```python
for path, key, description in [
    (
        annotation,
        "chromatin_accessibility/gencode_v41_GRCh38.gff3.gz",
        "GENCODE v41 gene annotation of GRCh38",
    ),
    (
        fasta,
        "chromatin_accessibility/gencode_v41_GRCh38.fa",
        "GENCODE v41 genome sequence of GRCh38",
    ),
    (
        motifs,
        "chromatin_accessibility/cisBP_human.meme",
        "human CIS-BP transcription factor motifs",
    ),
]:
    ln.Artifact(path, key=key, description=description).save()
```

```python
ln.finish()
```
