# توالی‌یابی RNA تک‌سلولی

این فصل، پرکاربردترین سنجش‌های توالی‌یابی ریبونوکلئیک اسید ({term}`RNA`) تک‌سلولی و مفاهیم پایهٔ مرتبط در زیست‌شناسی مولکولی را به‌اختصار معرفی می‌کند.
سنجش‌های {term}`چندوجهی <Modalities>` یا مکانی در اینجا بررسی نمی‌شوند، بلکه در فصل‌های پیشرفتهٔ مربوط معرفی خواهند شد.
هر سنجش {term}`توالی‌یابی <sequencing>` نقاط قوت و محدودیت‌های خاص خود را دارد؛ تحلیلگران داده باید آن‌ها را بشناسند تا از سوگیری‌های احتمالی داده آگاه باشند.

## واحد سازندهٔ حیات

حیات، چنان‌که می‌شناسیم، ویژگی‌ای است که موجودات زنده را از موجودات مرده یا اشیای بی‌جان متمایز می‌کند.
بیشتر تعریف‌های حیات یک جزء مشترک دارند: {term}`سلول‌ها <Cell>`.
سلول‌ها سامانه‌هایی باز هستند که هم‌ایستایی را حفظ می‌کنند، سوخت‌وساز دارند، رشد می‌کنند، با محیط سازگار می‌شوند، تولیدمثل می‌کنند، به محرک‌ها پاسخ می‌دهند و خود را سازمان می‌دهند.
بنابراین، سلول‌ها واحدهای بنیادی سازندهٔ حیات‌اند و نخستین بار در سال ۱۶۶۵ به‌دست دانشمند بریتانیایی Robert Hooke کشف شدند.
Hooke برش نازکی از چوب‌پنبه را با میکروسکوپی بسیار ابتدایی بررسی کرد و با شگفتی دید که ظاهر برش به کندوی عسل شباهت دارد.
او این واحدهای کوچک را «سلول» نامید.

در سال ۱۸۳۹، Matthias Jakob Schleiden و Theodor Schwann نخستین بار نظریهٔ سلولی را بیان کردند که بر اساس آن، همهٔ جانداران از سلول ساخته شده‌اند.
از زمان تعریف اولیهٔ نظریهٔ سلولی، پژوهشگران دریافته‌اند که همهٔ سلول‌ها تقریباً ترکیب شیمیایی یکسانی دارند و جریان پویایی از اطلاعات را نشان می‌دهند که رمز ژنتیکی را در قالب دئوکسی‌ریبونوکلئیک اسید ({term}`DNA`) از یک سلول به سلول دیگر منتقل می‌کند.
دو نوع کلی سلول وجود دارد: یوکاریوت و پروکاریوت.
سلول‌های یوکاریوتی هسته دارند که در آن، غشای هسته‌ای کروموزوم‌ها را در بر می‌گیرد؛ درحالی‌که سلول‌های پروکاریوتی فقط ناحیهٔ نوکلئوئید دارند و فاقد هسته‌اند.
هسته، DNA ژنومی سلول را در خود جای می‌دهد و نام یوکاریوت‌ها به همین ویژگی مربوط است: واژهٔ لاتینی _Nucleus_ به معنای مغز یا دانه است.
دستگاه همانندسازی DNA، اطلاعات وراثتی ذخیره‌شده در DNA هسته را می‌خواند تا آن را تکثیر کند و چرخهٔ حیات تداوم یابد.
DNA یوکاریوتی به چندین مجموعهٔ خطی به نام کروموزوم تقسیم شده است که هنگام تقسیم هسته با دوک میکروتوبولی از هم جدا می‌شوند.
درک اطلاعات وراثتی نهفته در DNA، کلید شناخت بسیاری از فرایندهای تکاملی و مرتبط با بیماری است.

توالی‌یابی فرایند تعیین ترتیب نوکلئوتیدهای DNA است.
کاربرد اصلی آن، آشکارکردن اطلاعات ژنتیکی حمل‌شده در یک قطعهٔ مشخص DNA، یک ژنوم کامل یا حتی میکروبیومی پیچیده است.
توالی‌یابی DNA به پژوهشگران امکان می‌دهد جایگاه، کارکرد و تنظیم یک ژن را شناسایی کنند. برای مثال، ویژگی‌های ژنتیکی مانند چارچوب‌های خوانش باز (ORF)، یعنی توالی‌های رمزکنندهٔ پروتئین میان کدون‌های شروع و توقف، یا جزایر {term}`CpG` را آشکار می‌کند که نشان‌دهندهٔ نواحی {term}`پروموتر <promoter>` هستند.
کاربرد گستردهٔ دیگر، تحلیل تکاملی است که در آن توالی‌های DNA همولوگ از جانداران مختلف مقایسه می‌شوند.
توالی‌یابی DNA همچنین برای بررسی ارتباط جهش‌ها با بیماری‌ها، یا گاهی حتی مقاومت به بیماری، به کار می‌رود؛ موضوعی که از ارزشمندترین کاربردهای آن محسوب می‌شود.

## تاریخچهٔ کوتاه توالی‌یابی

### توالی‌یابی نسل اول

اگرچه Friedrich Miescher نخستین بار DNA را در سال ۱۸۶۹ جداسازی کرد، بیش از ۱۰۰ سال طول کشید تا جامعهٔ علمی فناوری‌های توالی‌یابی با توان عملیاتی بالا را توسعه دهد.
در سال ۱۹۵۳، Watson، Crick و Franklin ساختار DNA را کشف کردند و در سال ۱۹۶۵، Robert Holley نخستین tRNA را توالی‌یابی کرد.
هفت سال بعد، در سال ۱۹۷۲، Walter Fiers نخستین کسی بود که یک ژن کامل، یعنی پروتئین پوششی باکتریوفاژ MS2، را توالی‌یابی کرد؛ او با RNaseها، RNA ویروس را هضم کرد، الیگونوکلئوتیدها را جداسازی کرد و در پایان آن‌ها را با الکتروفورز و کروماتوگرافی از هم جدا ساخت {cite:p}`scrna:Jou1972`.
هم‌زمان، Frederick Sanger روشی برای توالی‌یابی DNA با استفاده از قطعات دارای نشان پرتوزا و هضم‌شدهٔ جزئی توسعه داد که «روش خاتمهٔ زنجیره» نام گرفت و بیشتر با عنوان «توالی‌یابی سنگر» شناخته می‌شود.
اگرچه توالی‌یابی سنگر هنوز به کار می‌رود، محدودیت‌هایی مانند نبود خودکارسازی و زمان‌بر بودن داشته است.
در سال ۱۹۸۷، Leroy Hood و Michael Hunkapiller دستگاه ABI 370 را توسعه دادند که فرایند توالی‌یابی سنگر را خودکار می‌کند.
مهم‌ترین نوآوری آن، نشان‌دارکردن خودکار قطعات DNA با رنگ‌های فلورسانس به‌جای مولکول‌های پرتوزا بود.
این تغییر نه‌تنها اجرای روش را ایمن‌تر کرد، بلکه امکان تحلیل داده‌های به‌دست‌آمده با رایانه را نیز فراهم ساخت {cite:p}`scrna:Hood1987`.

```{dropdown} 👍 نقاط قوت
- توالی‌یابی سنگر ساده و مقرون‌به‌صرفه است.
- اگر درست انجام شود، نرخ خطای آن بسیار پایین است؛ کمتر از ۰٫۰۰۱٪.
```

```{dropdown} 👎 محدودیت‌ها
- روش‌های سنگر فقط می‌توانند قطعات کوتاه DNA با طول حدود ۳۰۰ تا ۱۰۰۰ جفت‌باز را توالی‌یابی کنند.
- کیفیت توالی سنگر اغلب در ۱۵ تا ۴۰ باز نخست پایین است، زیرا آغازگرها در این ناحیه متصل می‌شوند.
- کیفیت توالی‌یابی پس از ۷۰۰ تا ۹۰۰ باز افت می‌کند.
- اگر قطعهٔ DNA توالی‌یابی‌شده کلون شده باشد، ممکن است بخشی از توالی ناقل کلون‌سازی، یعنی حامل DNA برای کپی‌کردن، ذخیره و تکثیر ژن‌ها، وارد توالی نهایی شود.
- هزینهٔ توالی‌یابی سنگر به‌ازای هر باز، از توالی‌یابی نسل دوم یا سوم بیشتر است.
```

### توالی‌یابی نسل دوم

نه سال بعد، در سال ۱۹۹۶، Mostafa Ronaghi، Mathias Uhlen و Pȧl Nyŕen با توسعهٔ پیروسکوئنسینگ، انقلابی در توالی‌یابی DNA ایجاد کردند و آغازگر توالی‌یابی نسل دوم شدند.
توالی‌یابی نسل دوم، که با نام توالی‌یابی نسل جدید (NGS) نیز شناخته می‌شود، عمدتاً با خودکارسازی بیشتر در آزمایشگاه، استفاده از رایانه و کوچک‌مقیاس‌کردن واکنش‌ها ممکن شد.
پیروسکوئنسینگ، نور تولیدشده در جریان سنتز پیروفسفات هنگام توالی‌یابی را اندازه‌گیری می‌کند.
این فرایند معمولاً «توالی‌یابی از طریق سنتز» نیز نامیده می‌شود.
دو سال بعد، Shankar Balasubramanian و David Klenerman در شرکت Solexa، فرایند توالی‌یابی از طریق سنتز را برای روشی جدید مبتنی بر رنگ‌های فلورسانس توسعه دادند و سازگار کردند.
فناوری Solexa همچنین پایهٔ دستگاه‌های توالی‌یابی Illumina است که بر بازار امروز تسلط دارند.
دستگاه Roche 454 که در سال ۲۰۰۵ توسعه یافت، نخستین دستگاهی بود که کل فرایند پیروسکوئنسینگ را در یک ماشین واحد خودکار می‌کرد.
Life Technologies چندین سامانهٔ دیگر نیز معرفی کرد؛ از جمله SOLiD در سال ۲۰۰۷، سامانه‌ای مبتنی بر «توالی‌یابی از طریق اتصال»، و Ion Torrent در سال ۲۰۱۱ که یون‌های هیدروژن را هنگام سنتز DNA جدید آشکار می‌کند.
به‌طور کلی، توالی‌یابی از طریق سنتز شامل افزودن تک‌نوکلئوتیدها به رشتهٔ DNA در حال رشد و آشکارسازی هر افزوده‌شدن است.
در مقابل، توالی‌یابی از طریق اتصال، برای تعیین توالی بر آشکارسازی اتصال پروب‌های کوتاه DNA به قطعه متکی است.

```{dropdown} 👍 نقاط قوت
- توالی‌یابی نسل دوم اغلب از نظر مواد شیمیایی لازم، ارزان‌ترین گزینه است.
- حتی مقدار کم مادهٔ اولیه را می‌توان به‌عنوان ورودی استفاده کرد.
- حساسیت بالا برای آشکارسازی واریانت‌های کم‌فراوان و پوشش جامع ژنوم.
- ظرفیت بالا با چندگانه‌سازی نمونه‌ها.
- امکان توالی‌یابی هم‌زمان هزاران ژن.
```

```{dropdown} 👎 محدودیت‌ها
- دستگاه‌های توالی‌یابی گران‌اند و اغلب باید به‌صورت مشترک با همکاران استفاده شوند.
- دستگاه‌های نسل دوم بزرگ و ثابت‌اند و برای کار میدانی طراحی نشده‌اند.
- به‌طور کلی، توالی‌یابی نسل دوم قطعات توالی کوتاه فراوان، یا خوانش، تولید می‌کند که استفاده از آن‌ها برای ژنوم‌های جدید دشوار است.
- کیفیت نتیجهٔ توالی‌یابی به ژنوم مرجع وابسته است.
```

### توالی‌یابی نسل سوم

نسل سوم توالی‌یابی، که امروزه با عنوان توالی‌یابی نسل جدید نیز شناخته می‌شود، دو نوآوری به بازار عرضه کرده است.
نخست، توالی‌یابی با خوانش بلند که آشکارسازی قطعات نوکلئوتیدی بسیار بلندتر از قطعات تولیدشده با توالی‌یابی نسل دوم را ممکن می‌کند.
دستگاه‌های معمول خوانش کوتاه Illumina، بسته به مدل، قطعاتی به طول ۷۵ تا ۳۰۰ جفت‌باز تولید می‌کنند.
در توالی‌یابی نسل سوم، دستگاه‌ها می‌توانند ده‌ها هزار جفت‌باز را بخوانند.
این ویژگی به‌ویژه برای سرهم‌بندی ژنوم‌های جدیدی مهم است که ژنوم مرجع در دسترس ندارند.
دوم، امکان توالی‌یابی بلادرنگ، پیشرفت عمدهٔ دیگر نسل سوم است.
با ترکیب این قابلیت و دستگاه‌های قابل‌حملی که کوچک‌اند و برای انجام واکنش‌های شیمیایی به دستگاه‌های پیچیدهٔ دیگری نیاز ندارند، توالی‌یابی اکنون برای «کار میدانی» آماده است و می‌توان آن را حتی دور از امکانات آزمایشگاهی برای گردآوری نمونه‌ها به کار برد.

```{admonition} نکته‌ای دربارهٔ طول توالی
:class: note, dropdown
- ۱ جفت‌باز (bp)
- ۱ کیلو جفت‌باز (kb) = ۱٬۰۰۰ bp
- ۱ مگا جفت‌باز (Mb) = ۱٬۰۰۰٬۰۰۰ bp
- ۱ گیگا جفت‌باز (Gb) = ۱٬۰۰۰٬۰۰۰٬۰۰۰ bp
```

شرکت Pacific Biosciences، یا PacBio، در سال ۲۰۱۰ توالی‌یابی با موج‌بَر مُد صفر را معرفی کرد که از نانوحفره‌های حاوی یک DNA پلیمراز منفرد استفاده می‌کند.
این روش امکان مشاهدهٔ مستقیم افزوده‌شدن هر نوکلئوتید را با آشکارسازهای نصب‌شده زیر نانوحفره‌ها فراهم می‌کند.
هر نوع نوکلئوتید با یک رنگ فلورسانس اختصاصی نشان‌دار می‌شود که هنگام افزوده‌شدن، سیگنال فلورسانس ساطع می‌کند؛ سپس این سیگنال‌ها به‌عنوان خروجی خواندن توالی اندازه‌گیری می‌شوند.
خوانش‌های به‌دست‌آمده از دستگاه‌های PacBio معمولاً ۸ تا ۱۵ kb طول دارند و امکان رسیدن به ۷۰ kb نیز وجود دارد.

Oxford Nanopore Technologies نخستین بار دستگاه قابل‌حمل MinION را در سال ۲۰۱۲ معرفی کرد. MinION و نسل‌های بعدی آن، یعنی خانواده‌های GridION و PromethION، دستگاه‌های توالی‌یابی DNA/RNA هستند که توانایی تولید خوانش‌های بلندتر از ۲ Mb را دارند.
کتابخانه‌های تک‌سلولی نانوپور معمولاً رونوشت‌های cDNA به طول ۹۰۰ bp تا ۲٫۷ kb تولید می‌کنند.
نکتهٔ قابل‌توجه این است که دستگاه MinION آن‌قدر کوچک است که در کف دست جا می‌گیرد.
ایدهٔ دستگاه‌های Oxford Nanopore، آشکارسازی تغییر جریان الکتریکی هنگام عبور اسیدهای نوکلئیک از نانوحفره‌های پروتئینی است {cite:p}`scrna:Jain2016`.

```{dropdown} 👍 نقاط قوت
- خوانش‌های بلند، سرهم‌بندی ژنوم‌های بزرگ و جدید را ممکن می‌کنند.
- دستگاه‌ها قابل‌حمل‌اند و برای کار میدانی مناسب هستند.
- امکان آشکارسازی مستقیم تغییرات اپی‌ژنتیکی توالی‌های DNA و RNA.
- سرعت! دستگاه‌های توالی‌یابی نسل سوم سریع‌اند.
```

```{dropdown} 👎 محدودیت‌ها
- برخی دستگاه‌های نسل سوم، نرخ خطای بالاتری از دستگاه‌های نسل دوم دارند؛ برای مثال، فناوری جدید توالی‌یابی از طریق انبساط Roche می‌کوشد این مشکل را برطرف کند {cite:p}`scrna:Jain2016`
{cite:p}`scrna:roche`.
- معرف‌ها معمولاً از معرف‌های توالی‌یابی نسل دوم گران‌ترند.
```

## نمای کلی فرایند NGS

اگرچه فناوری‌های NGS متنوع‌اند، گام‌های عمومی توالی‌یابی DNA، و در نتیجه RNA رونوشت‌برداری معکوس‌شده، تا حد زیادی یکسان‌اند.
تفاوت‌ها عمدتاً در شیمی فناوری‌های توالی‌یابی مربوط است.

1. **آماده‌سازی نمونه و کتابخانه**: در گام نخست، با قطعه‌قطعه‌کردن نمونه‌های DNA و اتصال آن‌ها به {term}`مولکول‌های آداپتور <Adapter sequences>`، یک {term}`کتابخانه <library>` آماده می‌شود.
   این آداپتورها در هیبریداسیون قطعات کتابخانه به بستر نقش دارند و جایگاه اتصال آغازگر را تشکیل می‌دهند.

2. **تکثیر و توالی‌یابی**: در گام دوم، کتابخانه به مولکول‌های تک‌رشته‌ای تبدیل می‌شود.
   طی یک مرحلهٔ تکثیر، مانند واکنش زنجیره‌ای پلیمراز، خوشه‌هایی از مولکول‌های DNA ایجاد می‌شوند.
   همهٔ خوشه‌ها در یک نوبت توالی‌یابی، واکنش‌های منفرد خود را انجام می‌دهند.

3. **خروجی داده و تحلیل**: خروجی آزمایش توالی‌یابی به فناوری و شیمی آن وابسته است.
   برخی دستگاه‌ها سیگنال‌های فلورسانس تولید می‌کنند که در فایل‌های خروجی اختصاصی ذخیره می‌شوند.
   برخی دیگر ممکن است سیگنال‌های الکتریکی تولید کنند که در قالب‌های فایل متناظر ذخیره می‌شوند.
   به‌طور کلی، حجم داده‌های تولیدشده، یعنی داده‌های خام، بسیار زیاد است.
   چنین داده‌هایی به پردازشی پیچیده و سنگین از نظر محاسباتی نیاز دارند.
   این موضوع در فصل پردازش داده‌های خام بیشتر بررسی می‌شود.

## توالی‌یابی RNA

تا اینجا، توالی‌یابی را با این فرض ناگفته معرفی کرده‌ایم که DNA توالی‌یابی می‌شود.
اما دانستن توالی DNA یک جاندار و موقعیت عناصر تنظیمی آن، اطلاعات بسیار اندکی دربارهٔ فرایندهای پویا و لحظه‌ای سلول می‌دهد.
توالی‌یابی RNA ‏(RNA-Seq) به دانشمندان امکان می‌دهد تصویری از وضعیت سلول‌ها، بافت‌ها یا جانداران در زمان توالی‌یابی، در قالب نمایه‌های بیان ژن، به دست آورند.
از این اطلاعات می‌توان برای آشکارسازی تغییرات وضعیت بیماری در پاسخ به درمان‌ها، عوامل محیطی، ژنوتیپ‌ها و سایر شرایط آزمایشی استفاده کرد.

RNA-Seq عمدتاً از پروتکل‌های توالی‌یابی DNA پیروی می‌کند، اما یک مرحلهٔ رونوشت‌برداری معکوس دارد که در آن {term}`DNA مکمل (cDNA) <complementary DNA (cDNA)>` از الگوی RNA ساخته می‌شود.
توالی‌یابی نوین RNA، برخلاف روش‌هایی مانند سنجش‌های مبتنی بر ریزآرایه یا PCR کمی همراه با رونوشت‌برداری معکوس که برای هدف‌گیری اختصاصی نواحی موردنظر به طراحی پروب نیاز دارند، نمونه‌برداری بدون سوگیری از رونوشت‌ها را ممکن می‌کند.
[سنجش‌های مبتنی بر ریزآرایه](#seealso-dropdown) از پروب‌ها، یعنی توالی‌های مکمل، برای آشکارسازی توالی‌های خاص موردنظر، مانند ژن‌ها، استفاده می‌کنند.
[PCR کمی همراه با رونوشت‌برداری معکوس](#seealso-dropdown)، مقدار RNA هدف را با پایش تکثیر مولکول DNA مکمل، یا cDNA، طی PCR اندازه‌گیری می‌کند.

نمایه‌های بیان ژن به‌دست‌آمده، آشکارسازی ایزوفرم‌های ژنی، همجوشی‌های ژنی، واریانت‌های تک‌نوکلئوتیدی و بسیاری از ویژگی‌های جالب دیگر را نیز ممکن می‌کنند.
توالی‌یابی نوین RNA به دانش پیشین محدود نیست و امکان ثبت ویژگی‌های شناخته‌شده و جدید را فراهم می‌کند.
حاصل، مجموعه‌داده‌هایی غنی است که می‌توان از آن‌ها برای تحلیل اکتشافی داده استفاده کرد.

(introduction-scrna-seq-key-takeaway-2)=

## توالی‌یابی RNA تک‌سلولی

### توالی‌یابی توده‌ای در برابر توالی‌یابی RNA تک‌سلولی

RNA-Seq عمدتاً به دو صورت انجام می‌شود: یا RNA مخلوط‌شدهٔ سلول‌ها از منبع موردنظر توالی‌یابی می‌شود، یعنی {term}`توالی‌یابی توده‌ای <Bulk RNA sequencing>`، یا ترنسکریپتوم هر سلول به‌صورت جداگانه توالی‌یابی می‌شود، یعنی توالی‌یابی تک‌سلولی.
مخلوط‌کردن RNA همهٔ سلول‌ها در بیشتر موارد از RNA-Seq تک‌سلولی (scRNA-Seq)، که از نظر آزمایشگاهی پیچیده است، ارزان‌تر و ساده‌تر است.
RNA-Seq توده‌ای نمایه‌های بیان میانگین‌گیری‌شده بر سلول‌ها را تولید می‌کند که معمولاً تحلیل آسان‌تری دارند؛ اما بخشی از پیچیدگی، مانند ناهمگنی نمایهٔ بیان سلول‌ها، را نیز پنهان می‌کنند؛ پیچیدگی‌ای که ممکن است به پاسخ پرسش موردنظر کمک کند.
برخی داروها یا اغتشاش‌ها ممکن است فقط بر {term}`انواع سلولی <Cell type>` مشخص یا برهم‌کنش میان انواع سلولی اثر بگذارند.
برای مثال، در سرطان‌شناسی ممکن است سلول‌های توموری نادر و مقاوم به دارو باعث عود شوند؛ شناسایی این وضعیت با RNA-Seq توده‌ای ساده، حتی در سلول‌های کشت‌شده، دشوار است.

برای آشکارکردن چنین روابطی، بررسی بیان ژن در سطح تک‌سلولی ضروری است.
بااین‌حال، scRNA-Seq ملاحظات متعددی دارد.
نخست، آزمایش‌های تک‌سلولی معمولاً گران‌ترند و اجرای صحیح آن‌ها دشوارتر است.
دوم، {term}`تحلیل پایین‌دستی <downstream analysis>` به‌دلیل افزایش تفکیک‌پذیری پیچیده‌تر می‌شود و نتیجه‌گیری نادرست آسان‌تر خواهد بود.

آزمایش تک‌سلولی عموماً گام‌هایی مشابه RNA-Seq توده‌ای دارد، مطابق توضیح بالا، اما به سازگارسازی‌هایی نیاز دارد.
همانند توالی‌یابی توده‌ای، توالی‌یابی تک‌سلولی به لیز، رونوشت‌برداری معکوس، تکثیر و در نهایت توالی‌یابی نیاز دارد.
افزون بر این، توالی‌یابی تک‌سلولی مستلزم جداسازی سلول‌ها و تفکیک فیزیکی آن‌ها در محفظه‌های واکنش کوچک‌تر، یا نوع دیگری از برچسب‌گذاری سلولی است تا بعداً بتوان ترنسکریپتوم‌های به‌دست‌آمده را به سلول‌های منشأ نسبت داد.
بنابراین، تفاوت بیشتر سنجش‌های تک‌سلولی نیز در همین گام‌هاست: جداسازی تک‌سلول، تکثیر رونوشت و توالی‌یابی، بسته به دستگاه توالی‌یابی.
اما پیش از شرح جزئیات توالی‌یابی RNA تک‌سلولی، باید چالش‌های زیستی و فنی اندازه‌گیری {term}`mRNA <Messenger RNA (mRNA)>` در چنین تفکیک‌پذیری دقیقی را بشناسیم.

#### اندازه‌گیری «پیام‌رسان‌ها»

در هستهٔ scRNA-Seq یک پرسش بنیادی قرار دارد: **دقیقاً** چه چیزی را می‌خواهیم اندازه‌گیری کنیم؟
در آزمایش‌های RNA-seq، تمرکز ما بر کمّی‌سازی RNA پیام‌رسان، یا mRNA، درون سلول‌های منفرد است.
این مولکول، به تعبیر Brenner، Jacob و Meselson در سال ۱۹۶۱، «واسطه‌ای ناپایدار است که اطلاعات را از ژن‌ها به ریبوزوم‌ها برای سنتز پروتئین حمل می‌کند»؛ توصیفی که اصطلاح «پیام‌رسان» را پدید آورد {cite:p}`scrna:brenner1961unstable`.
بنابراین، mRNA حلقهٔ ارتباطی اساسی میان DNA و تولید پروتئین است.
بااین‌حال، mRNA فقط بخش کوچکی از کل RNA سلول را تشکیل می‌دهد.
حدود ۳ تا ۷ درصد جرم RNA متعلق به mRNA است؛ بخش عمدهٔ باقی‌مانده RNA غیررمزکننده است: ۸۰ تا ۹۰ درصد RNA ریبوزومی (rRNA)، ۱۰ تا ۱۵ درصد RNA ناقل (tRNA) و حدود ۱ درصد گونه‌های غیررمزکنندهٔ دیگر {cite:p}`scrna:palazzo2015non`؛ بنگرید به [مرور RNA غیررمزکننده](https://www.bio-rad.com/de-de/applications-technologies/coding-non-coding-rna?ID=Q1070M70KWE7).
برآوردها حاکی از وجود ۱۰۰ هزار تا یک میلیون مولکول mRNA در یک سلول معمول پستاندارند که تا حدود ۵۰ درصد همهٔ ژن‌ها را پوشش می‌دهند {cite:p}`scrna:velculescu1999analysis,scrna:Islam2014`.
این بدان معناست که در هر سلول معین، تعداد قابل‌توجهی از ژن‌ها اصلاً رونویسی نمی‌شوند؛ بازتابی از هویت و کارکرد اختصاصی آن سلول.
بااین‌حال، محدودیت‌های فنی فناوری‌های فعلی scRNA-seq اندازه‌گیری را پیچیده‌تر می‌کنند.
برای مثال، سامانه‌های پرکاربردی مانند 10X Genomics در هر نوبت فقط تا ۶۵ درصد سلول‌ها را به دام می‌اندازند و تنها حدود ۱۴ درصد mRNA هر سلول را بازیابی می‌کنند {cite:p}`scrna:aljanahi2018introduction`.
این محدودیت‌ها، آشکارسازی ژن‌های کم‌بیان را به‌ویژه دشوار می‌کنند.

نگاه به بیان ژن از دریچهٔ این اعداد، هم پیچیدگی زیستی و هم محدودیت ابزارهای ما را آشکار می‌کند؛ به [شکل کمّی‌سازی بیان ژن](#quantifying-gene-expression) مراجعه کنید.
برای درک عمیق‌تر، مسیر ژن تا پروتئین را گام‌به‌گام دنبال می‌کنیم.

#### از ژن تا پروتئین

مسیر ما با یک ژن آغاز می‌شود؛ ناحیه‌ای مشخص از DNA که الگوی سنتز mRNA است.
اگرچه تعداد ژن‌ها ممکن است میان افراد اندکی، در حدود ۷۰ ژن، تفاوت داشته باشد، ژنوم متوسط انسان حدود ۲۲ هزار ژن دارد {cite:p}`scrna:pertea2010between`.
رونویسی ژن به‌هیچ‌وجه پیوسته نیست.
در عوض، به‌صورت فوران‌های تصادفی رخ می‌دهد؛ دوره‌های کوتاه و نامنظم فعالیت که طی آن‌ها یک ژن ممکن است ناگهان چندین رونوشت mRNA تولید کند و سپس دوباره خاموش شود {cite:p}`scrna:suter2011mammalian`.
به همین دلیل، رونویسی mRNA را با توزیع دوجمله‌ای منفی مدل می‌کنیم.
این توزیع مناسب است، زیرا شمار رویدادها، یعنی mRNA، را مدل می‌کند و هم‌زمان بیش‌پراکندگی، یعنی بزرگ‌تر بودن واریانس از میانگین، را که بر اثر فوران‌های رونویسی ایجاد می‌شود در نظر می‌گیرد {cite:p}`scrna:ren2020negative,scrna:love2015deseq2`.

```{admonition} چرا توزیع دوجمله‌ای منفی؟
:class: dropdown

برای شمارش رویدادها در یک بازهٔ ثابت معمولاً از توزیع پواسون استفاده می‌شود؛ اما پواسون فقط یک پارامتر دارد و در نتیجه واریانس آن باید با میانگینش برابر باشد.
فوران رونویسی این فرض را نقض می‌کند: سلولی که هنگام اندازه‌گیری در میانهٔ یک فوران قرار دارد، رونوشت‌های بسیار بیشتری از یک ژن نسبت به سلولی در فاصلهٔ میان دو فوران دارد؛ بنابراین، پراکندگی شمارش‌ها از آنچه پواسون اجازه می‌دهد بیشتر است.

اگر نرخ پواسون خود اجازه داشته باشد از سلولی به سلول دیگر، مطابق توزیع گاما، تغییر کند، حاصل توزیع دوجمله‌ای منفی است.
این توزیع تفسیر شمارشی را حفظ می‌کند و یک پارامتر دوم، یعنی پراکندگی، به دست می‌آورد که این گسترش اضافی را در بر می‌گیرد.

ژنی را در نظر بگیرید که به‌طور میانگین ۵ شمارش در هر سلول تولید می‌کند.
در مدل پواسون، واریانس نیز ۵ است و تقریباً همهٔ سلول‌ها بین ۰ تا ۱۱ شمارش خواهند داشت.
در اندازه‌گیری سلول‌های واقعی، برای همین ژن معمولاً هم صفر دیده می‌شود و هم سلول‌هایی با ۳۰ شمارش یا بیشتر؛ زیرا فوران‌ها نرخ زیربنایی را در برخی سلول‌ها بالا و در برخی دیگر نزدیک صفر نگه داشته‌اند.
پارامتر پراکندگی به مدل اجازه می‌دهد هر دو حالت را مشاهده‌هایی معمول بداند، به‌جای آن‌که مقدار ۳۰ را دادهٔ پرت تلقی کند.
```

رونوشت اولیهٔ RNA، موسوم به پیش‌mRNA، سپس پیرایش جایگزین را می‌گذراند؛ فرایندی که امکان اتصال نواحی مختلف رونوشت، به نام اینترون‌ها و اگزون‌ها، را به شیوه‌های متعدد فراهم می‌کند.
این بدان معناست که یک ژن می‌تواند چندین ایزوفرم متمایز mRNA تولید کند.
به‌طور میانگین، هر ژن انسان حدود ۳٫۴ ایزوفرم mRNA تولید می‌کند {cite:p}`scrna:lee2015mechanisms`.
درحالی‌که همهٔ ژن‌های انسان دست‌کم دو ایزوفرم جایگزین دارند، برخی مرزهای پیچیدگی را فراتر می‌برند.
برای مثال، ژن basonuclin 2 انسان بالقوه می‌تواند تا ۹۰ هزار ایزوفرم mRNA تولید کند که به بیش از ۲۰۰۰ پروتئین متفاوت منجر می‌شوند {cite:p}`scrna:vanhoutteghem2007human`.
اما در برخی موارد، پیرایش جایگزین می‌تواند به آنزیمی فاقد کارکرد و ایجاد وضعیت بیماری نیز منجر شود.
در نهایت، این mRNA «بالغ» به پروتئین ترجمه می‌شود.
در این مرحله نیز اعداد به‌شدت متفاوت‌اند.
در پستانداران، میانهٔ نسبت پروتئین به mRNA حدود ۱۰ هزار پروتئین به‌ازای هر mRNA برآورد شده است {cite:p}`scrna:li2014system`.
بااین‌حال، بسته به ژن، نوع سلول و عوامل متعدد دیگر، این مقدار می‌تواند از چند صد تا نزدیک یک میلیون پروتئین به‌ازای هر رونوشت متغیر باشد {cite:p}`scrna:edfors2016gene`.
در نهایت، این فرایند به وجود حدود یک میلیارد پروتئین در یک سلول انسان منجر می‌شود {cite:p}`scrna:milo2013total`.

شناخت این لایه‌ها، از فوران‌های رونویسی و پیرایش جایگزین تا ترجمهٔ پروتئین، نشان می‌دهد که بیان ژن صرفاً مسیری ایستا نیست، بلکه سامانه‌ای پویا و احتمالاتی است.
اندازه‌گیری آن با تفکیک‌پذیری تک‌سلولی بینش عمیقی فراهم می‌کند، اما چالش‌ها و محدودیت‌های فناوری‌های فعلی ما را نیز آشکار می‌سازد.

### کمّی‌سازی رونوشت

کمّی‌سازی رونوشت، فرایند تبدیل داده‌های خام به جدولی از شمارش برآوردشدهٔ رونوشت برای هر ژن در هر نمونه، در توالی‌یابی توده‌ای، یا در هر سلول، در توالی‌یابی تک‌سلولی، است.
جزئیات بیشتر این فرایند محاسباتی در [فصل بعد](../introduction/raw_data_processing.md) شرح داده خواهد شد.

دو رویکرد عمده برای کمّی‌سازی رونوشت وجود دارد: تمام‌طول و مبتنی بر برچسب.
پروتکل‌های تمام‌طول می‌کوشند با خوانش‌های توالی‌یابی، کل رونوشت را به‌طور یکنواخت پوشش دهند؛ درحالی‌که پروتکل‌های مبتنی بر برچسب فقط انتهای 5' یا 3' را ثبت می‌کنند.
روش کمّی‌سازی رونوشت تأثیر مهمی بر ژن‌های ثبت‌شده دارد؛ بنابراین، تحلیلگران باید از فرایند کمّی‌سازی مورداستفاده آگاه باشند.
توالی‌یابی تمام‌طول به پروتکل‌های مبتنی بر پلیت محدود است؛ [به ادامه مراجعه کنید](#plate-based). آماده‌سازی کتابخانهٔ آن با رویکردهای RNA-seq توده‌ای قابل‌مقایسه است.
در پروتکل‌های تمام‌طول همیشه پوشش یکنواخت رونوشت‌ها حاصل نمی‌شود؛ بنابراین، نواحی مشخصی در طول بدنهٔ ژن همچنان ممکن است سوگیری داشته باشند.
یکی از مزیت‌های اصلی پروتکل‌های تمام‌طول، امکان آشکارسازی گونه‌های حاصل از پیرایش است.

پروتکل‌های مبتنی بر برچسب فقط انتهای 3' یا 5' رونوشت‌ها را توالی‌یابی می‌کنند.
بهای این کار، پوشش‌ندادنِ لزوماً کامل طول ژن است که هم‌ترازی بی‌ابهام خوانش‌ها با یک رونوشت و تمایز ایزوفرم‌های مختلف را دشوار می‌کند {cite:p}`scrna:Archer2016`.
بااین‌حال، این روش امکان استفاده از شناسه‌های مولکولی یکتا ({term}`UMIها <Unique Molecular Identifier (UMI)>`) را فراهم می‌کند که برای رفع سوگیری‌های فرایند تکثیر رونوشت سودمندند.

تکثیر رونوشت مرحله‌ای حیاتی در هر نوبت RNA-seq است تا فراوانی رونوشت‌ها برای کنترل کیفیت و توالی‌یابی کافی باشد.
در این فرایند، که معمولاً با واکنش زنجیره‌ای پلیمراز ({term}`PCR`) انجام می‌شود، از قطعات یکسان مولکول اصلی، کپی ساخته می‌شود.
ازآنجاکه کپی‌ها و مولکول‌های اصلی از هم قابل‌تشخیص نیستند، تعیین تعداد اولیهٔ مولکول‌ها در نمونه‌ها دشوار می‌شود.
UMIها راه‌حلی رایج برای کمّی‌سازی مولکول‌های اصلی و غیرتکراری هستند.

UMIها به‌عنوان {term}`بارکدهای <barcode>` مولکولی عمل می‌کنند و گاهی بارکد تصادفی نامیده می‌شوند.
این «بارکدها» از توالی‌های کوتاه و تصادفی نوکلئوتیدی تشکیل شده‌اند که به‌عنوان برچسب یکتا به هر مولکول نمونه افزوده می‌شوند.
UMI باید هنگام تولید کتابخانه و پیش از مرحلهٔ تکثیر اضافه شود.
شناسایی دقیق موارد تکراری PCR برای تحلیل پایین‌دستی اهمیت دارد تا {term}`سوگیری‌های تکثیر <amplification bias>` کنار گذاشته شوند یا دست‌کم از آن‌ها آگاه باشیم {cite:p}`scrna:Aird2011`.

سوگیری تکثیر به توالی‌های RNA/cDNA اشاره دارد که ترجیحاً تکثیر می‌شوند و در نتیجه بیشتر توالی‌یابی می‌شوند و شمارش بالاتری به دست می‌آورند.
این پدیده می‌تواند به هر تحلیل بیان ژن آسیب بزند، زیرا ژن‌هایی که چندان فعال نیستند ممکن است ناگهان پُربیان به نظر برسند.
این مسئله به‌ویژه دربارهٔ توالی‌هایی صادق است که در مراحل دیرتر PCR تکثیر می‌شوند؛ جایی که نرخ خطا ممکن است در مقایسه با مراحل اولیه بیشتر شده باشد.
اگرچه می‌توان از نظر محاسباتی چنین توالی‌هایی را با حذف خوانش‌های دارای مختصات هم‌ترازی یکسان شناسایی و حذف کرد، توصیهٔ عمومی این است که در صورت امکان، آزمایش همواره با UMI طراحی شود.
استفاده از UMI همچنین نرمال‌سازی شمارش ژن را بدون افت دقت ممکن می‌کند {cite:p}`scrna:Kivioja2012`.

### پروتکل‌های توالی‌یابی تک‌سلولی

پروتکل‌های فراوانی برای توالی‌یابی ترنسکریپتوم سلول‌های منفرد وجود دارد.
بااین‌حال، اصطلاحات به‌ویژه برای تازه‌واردان به این حوزه، اغلب می‌توانند مبهم باشند.
برای روشن‌شدن موضوع، این روش‌ها را بر اساس نحوهٔ جداسازی سلول‌ها در دو گروه کلی قرار می‌دهیم:

- **جداسازی در قطره‌ها**: این روش‌ها سلول‌های منفرد را در قطره‌های کوچک درون یک امولسیون محصور می‌کنند و پردازش با توان عملیاتی بالا را ممکن می‌سازند.
- **جداسازی در محفظه‌های فیزیکی**: این روش‌ها سلول‌ها را در محفظه‌های فیزیکی مجزا، که اغلب چاهک نامیده می‌شوند، جدا می‌کنند.

هر رویکرد از نظر توانایی بازیابی رونوشت‌ها، تعداد سلول‌های توالی‌یابی‌شده و بسیاری جنبه‌های دیگر متفاوت است.
در زیربخش‌های بعد، به‌اختصار نحوهٔ عملکرد، نقاط قوت و ضعف و سوگیری‌های احتمالی هر پروتکل را که تحلیلگران داده باید بشناسند بررسی می‌کنیم.

#### جداسازی در قطره‌ها

##### رایج‌ترین پروتکل‌ها

پرکاربردترین پروتکل‌ها **inDrop** {cite:p}`scrna:Klein2015`، **Drop-seq** {cite:p}`exp:Macosko2015` و سامانهٔ تجاری **10x Genomics Chromium** {cite:p}`exp:Zheng2017` هستند.
این پروتکل‌ها از ریزسیال‌ها برای به‌دام‌اندازی سلول‌ها در حباب‌های کوچک آبی به نام قطره استفاده می‌کنند.
هر قطره فضایی جداگانه تشکیل می‌دهد که فقط یک سلول و مواد شیمیایی لازم، یعنی مهره‌ها، را در بر دارد.
پروتکل‌های یادشده می‌توانند در هر ثانیه هزاران قطره تولید کنند.
این فرایند به‌شدت موازی، تعداد بسیار زیادی قطره با هزینه‌ای نسبتاً کم تولید می‌کند.

```{admonition} تولید قطره با ورتکس‌کردن
:class: dropdown, note
پروتکل PIP-seq جایگزینی ساده‌تر برای روش‌های مرسوم ریزسیالی در تولید قطرات آب در روغنِ هم‌اندازه ارائه می‌دهد.
برخلاف دستگاه‌های پیچیدهٔ ریزسیالی که به تجهیزات و مهارت تخصصی نیاز دارند، PIP-seq با ورتکس‌کردن سادهٔ محلول، قطره تولید می‌کند.
این روش را می‌توان به‌سادگی با افزایش حجم ظرف مقیاس‌پذیر کرد، بدون آن‌که زمان امولسیون‌سازی، که محدودیتی رایج در ریزسیال‌هاست، مانع شود {cite:p}`scrna:clark2023microfluidics`.

بااین‌حال، با وجود سادگی، {term}`ارزیابی‌های مقایسه‌ای <benchmark>` مستقل نشان می‌دهند PIP-seq در مقایسه با روش‌های تثبیت‌شده همچنان محدودیت‌هایی دارد.
برای مثال، PIP-seq به حدود ۱۵۰۰ شمارش ژنی رسید، درحالی‌که بهترین کیت 10x Genomics Chromium حدود ۴۰۰۰ شمارش ژنی نشان داد {cite:p}`scrna:de2025comprehensive`.
این یافته‌ها بیانگر موازنهٔ میان آسانی استفاده و عملکرد در نسخهٔ فعلی پروتکل PIP-seq هستند.
```

اگرچه این سه پروتکل در جزئیات متفاوت‌اند، قطرات نانولیتری حاوی سلول‌های محصورشده همواره برای به‌دام‌اندازی هم‌زمان مهره و سلول طراحی می‌شوند.
در فرایند محصورسازی از ریزمهره‌های اختصاصی با آغازگرهای متصل به مهره استفاده می‌شود که شامل دستگیرهٔ PCR، بارکد سلول، UMI به طول ۴ تا ۸ جفت‌باز و دُم poly-T هستند؛ در کیت 5'، آغازگر poly-T وجود خواهد داشت.
پس از لیز، mRNA سلول بی‌درنگ آزاد و به‌وسیلهٔ الیگونوکلئوتیدهای بارکددار متصل به مهره‌ها به دام انداخته می‌شود.
سپس قطره‌ها گردآوری و شکسته می‌شوند تا ترنسکریپتوم‌های تک‌سلولی متصل به ریزذرات، موسوم به STAMPs، آزاد شوند.
پس از آن، PCR و رونوشت‌برداری معکوس برای ثبت و تکثیر رونوشت‌ها انجام می‌شود.
در پایان، تگمنتاسیون رخ می‌دهد که طی آن رونوشت‌ها به‌طور تصادفی بریده می‌شوند و آداپتورهای توالی‌یابی به آن‌ها متصل می‌شوند.
این فرایند کتابخانه‌هایی آمادهٔ توالی‌یابی تولید می‌کند، مطابق توضیح پیشین.
در پروتکل‌های مبتنی بر قطره، فقط حدود ۱۰ درصد رونوشت‌های سلول بازیابی می‌شوند {cite:p}`scrna:Islam2014`.
نکتهٔ قابل‌توجه آن است که همین میزان کم توالی‌یابی برای شناسایی مقاوم انواع سلولی کافی است.

هر سه روش، سوگیری‌های مشخصی ایجاد می‌کنند.
جنس مهره‌های مورداستفاده در پروتکل‌ها متفاوت است.
در {term}`Drop-seq`، مهره از رزین شکننده ساخته می‌شود.
بنابراین، محصورسازی مهره‌ها از {term}`توزیع پواسون <Poisson distribution>` پیروی می‌کند؛ درحالی‌که مهره‌های {term}`InDrop` و 10X Genomics تغییرشکل‌پذیرند و در نتیجه نرخ اشغال با مهره به بیش از ۸۰ درصد می‌رسد {cite:p}`scrna:Zhang2019`.

افزون بر این، استفاده از آغازگرهای متصل به سطح در Drop-Seq احتمالاً بر کارایی به‌دام‌اندازی اثر می‌گذارد.
InDrop از آغازگرهایی استفاده می‌کند که با برش نوری آزاد می‌شوند و 10X genomics مهره‌ها را حل می‌کند.
این تفاوت بر محل انجام رونوشت‌برداری معکوس نیز اثر دارد.
در Drop-seq، رونوشت‌برداری معکوس پس از آزادشدن مهره‌ها از قطرات انجام می‌شود؛ درحالی‌که در InDrop و 10X genomics، این فرایند درون قطره رخ می‌دهد {cite:p}`scrna:Zhang2019`.

مقایسهٔ Zhang و همکاران در سال ۲۰۱۹ نشان داد که 10X Genomics از نظر کیفیت مهره از inDrop و Drop-seq بهتر عمل می‌کند، زیرا بارکدهای سلولی در دو سامانهٔ اخیر، عدم‌تطابق‌های آشکاری داشتند.
افزون بر این، سهم خوانش‌های منشأگرفته از بارکدهای معتبر برای 10X Genomics برابر ۷۵ درصد بود، درحالی‌که برای InDrop فقط ۲۵ درصد و برای Drop-seq برابر ۳۰ درصد بود.

مزیت‌های مشابهی از نظر حساسیت برای 10X Genomics نشان داده شد.
در این مقایسه، 10X Genomics به‌طور میانگین حدود ۱۷ هزار رونوشت از ۳۰۰۰ ژن را ثبت کرد؛ در مقایسه با ۸۰۰۰ رونوشت از ۲۵۰۰ ژن برای Drop-seq و ۲۷۰۰ رونوشت از ۱۲۵۰ ژن برای InDrop.
نویز فنی در 10X Genomics کمترین مقدار را داشت و پس از آن Drop-seq و InDrop قرار داشتند {cite:p}`scrna:Zhang2019`.

داده‌های تولیدشده، سوگیری‌های بزرگ وابسته به پروتکل را نشان دادند.
10X Genomics به‌دام‌اندازی و تکثیر ژن‌های کوتاه‌تر و دارای محتوای GC بالاتر را ترجیح می‌داد؛ درحالی‌که Drop-seq در مقایسه، ژن‌هایی با محتوای GC پایین‌تر را بیشتر ثبت می‌کرد.
اگرچه نشان داده شد 10X Genomics از جنبه‌های مختلف از سایر پروتکل‌ها بهتر عمل می‌کند، هزینهٔ آن به‌ازای هر سلول نیز حدود دو برابر است.
افزون بر این، Drop-seq، به‌جز مهره‌های آن، متن‌باز است و در صورت نیاز می‌توان پروتکل را آسان‌تر سازگار کرد.
InDrop کاملاً متن‌باز است و حتی مهره‌های آن را می‌توان در آزمایشگاه ساخت و اصلاح کرد.
بنابراین، InDrop منعطف‌ترین گزینه در میان این سه پروتکل است.

```{dropdown} 👍 نقاط قوت
- توالی‌یابی تعداد زیادی سلول را با هزینهٔ مناسب برای شناسایی ترکیب کلی بافت و توصیف انواع سلولی نادر ممکن می‌کند.
- می‌توان UMI را در آن گنجاند.
```

```{dropdown} 👎 محدودیت‌ها
- نرخ آشکارسازی رونوشت در مقایسه با روش‌های دیگر پایین است.
- فقط انتهای 3'، یا بسته به کیت انتهای 5'، ثبت می‌شود و رونوشت کامل ثبت نمی‌شود.
```

##### پیوند توالی‌یابی نانوپور و فناوری قطره

رویکردهای توالی‌یابی تک‌سلولی با خوانش بلند، به‌ندرت از UMI استفاده می‌کنند {cite:p}`scrna:Singh2019` یا تصحیح UMI انجام نمی‌دهند {cite:p}`scrna:Gupta2018` و در نتیجه برخی خوانش‌ها را به‌اشتباه به UMIهای جدید نسبت می‌دهند.
به‌دلیل نرخ خطای بالاتر دستگاه‌های خوانش بلند، این موضوع مشکلات جدی ایجاد می‌کند {cite:p}`scrna:Lebrigand2020`.
Lebrigand و همکاران ScNaUmi-seq، مخفف Single-cell Nanopore sequencing with UMIs، را معرفی کردند که توالی‌یابی نانوپور را با تخصیص بارکد سلولی و UMI ترکیب می‌کند.
تخصیص بارکد با داده‌های Illumina هدایت می‌شود؛ توالی بارکدهای سلولی یافت‌شده در خوانش‌های نانوپور با بارکدهای بازیابی‌شده از خوانش‌های Illumina برای همان ناحیه یا ژن مقایسه می‌شوند {cite:p}`scrna:Lebrigand2020`.
اما این کار عملاً به دو کتابخانهٔ تک‌سلولی نیاز دارد.
ScCOLOR-seq با استفاده از جفت‌های نوکلئوتیدی مکمل در سراسر طول بارکد، بارکدهای بدون خطا را به‌صورت محاسباتی شناسایی می‌کند.
سپس این بارکدها به‌عنوان راهنما برای تصحیح بارکدهای خطادار باقی‌مانده به کار می‌روند {cite:p}`scrna:Philpott2021`.
نسخه‌ای اصلاح‌شده از روش مبتنی بر شبکهٔ جهت‌دار UMI-tools، تکرار توالی‌های UMI را تصحیح می‌کند.

```{dropdown} 👍 نقاط قوت
- اطلاعات ناهمگنی پیرایش و توالی را بازیابی می‌کند.
```

```{dropdown} 👎 محدودیت‌ها
- معرف‌های نانوپور گران‌اند.
- نرخ خطای بازیابی بارکد سلولی بالاست.
- بسته به پروتکل، تخصیص بارکد با داده‌های Illumina هدایت می‌شود و به دو سنجش توالی‌یابی نیاز دارد.
```

#### جداسازی در محفظه‌های فیزیکی

(plate-based)=

##### پروتکل‌های مبتنی بر پلیت

پروتکل‌های مبتنی بر پلیت معمولاً سلول‌ها را به‌صورت فیزیکی در پلیت‌های دارای ریزچاهک جدا می‌کنند.
گام نخست، مرتب‌سازی سلول‌ها، برای مثال با جداسازی سلولی فعال‌شونده با فلورسانس (FACS)، است که در آن سلول‌ها بر اساس نشانگرهای سطحی مشخص جدا می‌شوند؛ یا این کار با میکروپیپتاژ انجام می‌شود.
سپس سلول‌های انتخاب‌شده در چاهک‌های منفرد حاوی بافر لیز سلولی قرار می‌گیرند.
در این چاهک‌ها رونوشت‌برداری معکوس انجام می‌شود.
به‌این‌ترتیب، در یک آزمایش می‌توان چند صد سلول را تحلیل کرد که از هر یک ۵۰۰۰ تا ۱۰ هزار ژن ثبت شده است.

پروتکل‌های توالی‌یابی مبتنی بر پلیت شامل SMART-seq2، MARS-seq، QUARTZ-seq و SRCB-seq هستند، اما به این‌ها محدود نمی‌شوند.
به‌طور کلی، این پروتکل‌ها از نظر توانایی چندگانه‌سازی متفاوت‌اند.
برای مثال، MARS-seq سه سطح بارکد، یعنی برچسب‌های مولکولی، سلولی و در سطح پلیت، را برای چندگانه‌سازی مقاوم فراهم می‌کند.
در مقابل، SMART-seq2 امکان چندگانه‌سازی زودهنگام ندارد و در نتیجه تعداد سلول‌ها را محدود می‌کند.
مقایسهٔ نظام‌مند پروتکل‌ها به‌دست Mereu و همکاران در سال ۲۰۲۰ نشان داد که QUARTZ-seq2 می‌تواند در هر سلول ژن‌های بیشتری از SMART-seq2، MARS-seq یا SRCB-seq ثبت کند {cite:p}`scrna:Mereu2020`.
این بدان معناست که QUARTZ-seq2 ژن‌های نشانگر اختصاصی نوع سلول را به‌خوبی ثبت می‌کند و برچسب‌گذاری مطمئن نوع سلول را ممکن می‌سازد.

```{dropdown} 👍 نقاط قوت
- ژن‌های فراوانی را در هر سلول بازیابی می‌کند و توصیف عمیق را ممکن می‌سازد.
- می‌توان پیش از آماده‌سازی کتابخانه، اطلاعاتی گردآوری کرد؛ برای مثال، با مرتب‌سازی FACS می‌توان اطلاعاتی مانند اندازهٔ سلول و شدت هر نشان به‌کاررفته را به مختصات چاهک مرتبط کرد.
- بازیابی رونوشت تمام‌طول را ممکن می‌کند.
```

```{dropdown} 👎 محدودیت‌ها

- مرحلهٔ قطعه‌قطعه‌سازی، اطلاعات اختصاصی رشته را از بین می‌برد {cite:p}`scrna:Hrdlickova2017`.
```

##### Fluidigm C1

سامانهٔ تجاری Fluidigm C1 تراشه‌ای ریزسیالی است که سلول‌ها را به‌طور خودکار بارگذاری و در محفظه‌های کوچک واکنش جدا می‌کند.
پروتکل‌های CEL-seq2 و SMART-seq، نسخهٔ ۱، در گردش‌کار خود از تراشه‌های Fluidigm C1 استفاده می‌کنند و امکان انجام مشترک استخراج RNA و آماده‌سازی کتابخانه را فراهم می‌سازند؛ در نتیجه، کار دستی لازم کاهش می‌یابد.
بااین‌حال، Fluidigm C1 به مخلوط‌های سلولی نسبتاً همگن نیاز دارد، زیرا سلول‌ها بر اساس اندازه به محل‌های متفاوت تراشهٔ ریزسیالی می‌رسند و این امر می‌تواند سوگیری مکانی ایجاد کند.
توالی‌یابی تمام‌طول ممکن است، زیرا مرحلهٔ تکثیر در چاهک‌های منفرد انجام می‌شود و در نتیجه سوگیری 3' بسیاری از پروتکل‌های دیگر RNA-seq تک‌سلولی عملاً کاهش می‌یابد.
این پروتکل به‌طور کلی گران‌تر نیز هست؛ بنابراین، بیشتر برای بررسی گستردهٔ یک جمعیت سلولی مشخص سودمند است.


```{dropdown} 👎 محدودیت‌ها
- فقط توالی‌یابی حداکثر ۸۰۰ سلول را ممکن می‌کند {cite:p}`scrna:fluidigm`.
- هزینهٔ هر سلول از پروتکل‌های دیگر بیشتر است.
- فقط حدود ۱۰ درصد سلول‌های استخراج‌شده به دام می‌افتند؛ بنابراین، این پروتکل برای انواع سلولی نادر یا ورودی کم مناسب نیست.
- آرایه‌های مورداستفاده فقط سلول‌هایی با اندازه‌های مشخص را ثبت می‌کنند و ممکن است در رونوشت‌های ثبت‌شده سوگیری ایجاد کنند.
```

افزون بر این، اگر بودجه عامل محدودکننده است، پروتکل منتخب باید مقرون‌به‌صرفه‌تر و مقاوم باشد.

### تک‌سلولی در برابر تک‌هسته‌ای

تا اینجا فقط سنجش‌های تک‌سلولی را بررسی کرده‌ایم، اما می‌توان تنها هستهٔ سلول‌ها را نیز توالی‌یابی کرد.
تعیین نمایهٔ تک‌سلولی همیشه تصویری بدون سوگیری از انواع سلولی در بافت‌ها یا اندام‌های مشخص، مانند مغز، ارائه نمی‌دهد.
طی فرایند تفکیک بافت، برخی انواع سلولی آسیب‌پذیرترند و در نتیجه به‌دام‌اندازی آن‌ها دشوار است. برای مثال، در نئوکورتکس موش، نورون‌های بینابینی سریع‌شلیک و پاروالبومین‌مثبت و نورون‌های گلوتاماترژیک با فرافکنی زیرقشری، با نسبتی کمتر از انتظار مشاهده شدند {cite:p}`scrna:Tasic2018`.
در مقابل، سلول‌های غیرنورونی بهتر از نورون‌ها تفکیک بافت را تحمل می‌کنند و در سوسپانسیون‌های تک‌سلولی نئوکورتکس انسان بالغ، بیش از نسبت واقعی بازنمایی می‌شوند {cite:p}`scrna:darmanis2015`.
افزون بر این، توالی‌یابی تک‌سلولی به‌شدت به بافت تازه متکی است و استفاده از زیست‌بانک‌های بافتی را دشوار می‌کند.
از سوی دیگر، هسته‌ها در برابر نیروی مکانیکی مقاوم‌ترند و می‌توان آن‌ها را بدون استفاده از آنزیم‌های تفکیک بافت، به‌طور ایمن از بافت منجمد جدا کرد {cite:p}`scrna:Krishnaswami2016`.
کاربردپذیری هر دو گزینه در بافت‌ها و انواع نمونه متفاوت است و سوگیری‌ها و عدم‌قطعیت‌های حاصل هنوز کاملاً روشن نشده‌اند.
پیش‌تر نشان داده شده است که هسته‌ها همهٔ الگوهای رونویسی سلول‌ها را به‌دقت بازتاب می‌دهند {cite:p}`scrna:Ding2020`.
انتخاب روش تک‌سلولی یا تک‌هسته‌ای در طراحی آزمایش، عمدتاً به نوع نمونهٔ بافتی بستگی دارد.
اما در تحلیل داده باید توجه داشت که قابلیت تفکیک بافت، اثر عمده‌ای بر انواع سلولی بالقوه قابل‌مشاهده دارد.
