<!-- END DROPDOWNS -->


(chromatin-accessibility-introduction-key-takeaway-1)=
## Motivation

Every cell of an organism carries the same genome, yet cells differ in which genes they express.
Gene expression is controlled by DNA methylation, histone modifications and the binding of transcription factors, and {term}`chromatin <Chromatin>` accessibility reflects the combined regulatory state that results from them.
Accessibility is therefore an orthogonal layer of information on cell identity, and it points to the regulatory elements and transcription factors that control gene expression.

:::{figure} ../_static/images/sc_atac/mechanisms_overview.png
:name: Mechanisms influencing chromatin accessibility
:alt: Accessibility regulation
:class: bg-primary mb-1
:width: 800px

Overview of mechanisms influencing chromatin accessibility. Created with BioRender.com.
:::

Accessibility is shaped at several scales, from the attachment of chromatin to the nuclear periphery {cite:p}`atac:narwade_mapping_2019, atac:buchwalter_coaching_2019` over chromatin remodelers and histone modifications to transcription factors that displace nucleosomes and the methylation of {term}`CpG` sites.
For a review of how these mechanisms regulate the genome, we refer to {cite:t}`atac:isbel_generating_2022`.

## Experimental assay

The Assay for Transposase-Accessible Chromatin with sequencing (ATAC-seq) uses the Tn5 transposase, which inserts sequencing adapters into accessible chromatin and thereby cuts it into fragments.
Only fragments flanked by two different adapters are amplified, sequenced paired-end and aligned to the genome.
In single-cell protocols such as 10x scATAC-seq, the transposition is performed on isolated nuclei in bulk before each nucleus is barcoded in a droplet.
The 10x Multiome assay additionally barcodes the RNA of each nucleus, which measures chromatin accessibility and gene expression in the same cells.
Nuclei are profiled instead of whole cells, so the RNA part of Multiome data measures nuclear transcripts.

(chromatin-accessibility-introduction-key-takeaway-2)=
## Data characteristics

Accessibility is measured across the whole genome, including non-coding regions, so there are no predefined features like genes in scRNA-seq.
Fragments are counted either in regions of high accessibility, called peaks, or in fixed-width tiles across the genome, which do not require peaks to be called first and cluster cells equally well {cite:p}`atac:luo2024benchmarking`.

A diploid cell has only two copies of each position, and only 1-10% of accessible regions are detected per cell, which makes the data very sparse {cite:p}`atac:chen_assessment_2019`.
Many methods therefore binarize counts, but treating counts quantitatively improves model fit, clustering, cell type identification and batch integration {cite:p}`atac:martens2024`.
Counting reads counts a fragment twice if both of its ends fall into a region but once otherwise, so fragments or pairs of insertions should be counted instead {cite:p}`atac:martens2024,atac:miao2024`.

(chromatin-accessibility-introduction-key-takeaway-3)=
## Analysis workflow

The main frameworks for scATAC-seq analysis are ArchR {cite:p}`atac:granja_archr_2021` and Signac {cite:p}`atac:stuart2021signac` in R, and SnapATAC2 {cite:p}`atac:zhang2024snapatac2` in Python.
In a benchmark of these pipelines, SnapATAC2 was among the best at separating cell types, particularly in datasets with complex cell type structures, and it was the fastest, which makes it our choice for this section {cite:p}`atac:luo2024benchmarking`.
Multimodal data such as Multiome can be stored as MuData objects and analyzed with muon {cite:p}`atac:bredikhin2022muon`, as introduced in the [chapter on advanced data structures](../introduction/advanced_data_structures_and_frameworks.ipynb).

The following chapters analyze the chromatin accessibility of about 10,000 PBMCs measured with the 10x Multiome assay, from [quality control](quality_control.ipynb) over [dimensionality reduction and clustering](dimensionality_reduction_clustering.ipynb) and [peak calling and motif analysis](peaks_motifs.ipynb) to [gene regulatory networks](gene_regulatory_networks_atac.ipynb) that combine accessibility and expression.

## Contributors

We gratefully acknowledge the contributions of:

### Authors

* Christopher Lance
* Laura Martens

### Reviewers

* Lukas Heumos
